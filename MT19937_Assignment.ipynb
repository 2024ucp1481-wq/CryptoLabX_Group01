# Assignment 10: MT19937 Stream Cipher and Predictable Token Attack

## Objective
To understand the weakness of statistical PRNGs in cryptography by implementing an MT19937 stream cipher, recovering a 16-bit seed using known plaintext, and predicting timestamp-based password reset tokens.

## 1. MT19937 Algorithm

MT19937 (Mersenne Twister) is a pseudorandom number generator.

- It uses an internal state of 624 words.
- Its period is 2^19937 - 1.
- It generates pseudorandom numbers using initialization, twisting and tempering operations.
- It is fast and provides good statistical randomness.

### Why MT19937 is not cryptographically secure

- Its output is predictable.
- Its internal state can be recovered from sufficient output.
- It does not provide cryptographic security.
- Predictable seeds make its output easier to guess.

Therefore, it should not be used for password reset tokens or encryption keys.

## 2. Known Plaintext Attack

A stream cipher uses XOR operation:

Ciphertext = Plaintext XOR Keystream

Decryption:

Plaintext = Ciphertext XOR Keystream

If plaintext is known, keystream can be recovered:

Keystream = Ciphertext XOR Plaintext

In this experiment, plaintext contains 14 consecutive A characters.

The seed is 16-bit.

Search Space = 2^16 = 65,536 possible seeds.

All possible seeds are tested to recover the correct seed.

## 3. Predictable Password Reset Tokens

The application generates a 16-byte token using MT19937.

The seed is the current Unix timestamp.

An attacker who knows the approximate generation time can test timestamps within a given interval.

Search Interval = T - 300 to T + 300

Total possible timestamps = 601

The matching timestamp is recovered by comparing generated tokens with the observed token.

PART A: SEED RECOVERY

1. Generate plaintext with random prefix and 14 A's.
2. Select a random 16-bit seed.
3. Encrypt plaintext using MT19937.
4. For every seed from 0 to 65535:
      Generate keystream.
      Decrypt ciphertext.
      Check for 14 consecutive A's.
5. Recover matching seed and plaintext.
6. Verify complete plaintext.


PART B: TOKEN PREDICTION

1. Record current Unix timestamp T.
2. Seed MT19937 with T.
3. Generate 16-byte token.
4. Define search interval T-300 to T+300.
5. Test each timestamp as a seed.
6. Compare generated token with observed token.
7. Recover matching timestamp.
8. Regenerate token and verify.
